```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from gdriver.gdriver import getCSVtoPandas, get_credentials
```

```python
# Get the data
creds = get_credentials()
st110_data = getCSVtoPandas('TonyGrove/LR_TG_C_AirTemp_ST110_Avg_SourceID_1_QC_1.csv',creds,skiprows=56)
```

```python
# Change index to LocalDateTime
st110_data['LocalDateTime'] = pd.to_datetime(st110_data['LocalDateTime'])
st110_data = st110_data.set_index('LocalDateTime')
```

```python
# Convert -9999 to NaN
st110_data = st110_data.replace(-9999, np.nan)
```
