## NLDAS download testing

```python
import requests
```

```python
# Define the base URL for the NLDAS Data Rods API
base_url = 'https://hydro1.gesdisc.eosdis.nasa.gov/daac-bin/access/timeseries.cgi'

# Define the parameters for the API request
params = {
    'variable': 'NLDAS2:NLDAS_FORA0125_H_v2.0:Tair',  # Example variable: 2-meter temperature
    'startDate': '2020-01-01T00',  # Start date in YYYY-MM-DDTHH format
    'endDate': '2020-01-02T00',    # End date in YYYY-MM-DDTHH format
    'location': 'GEOM:POINT(-97.0, 38.0)',  # Location as a point (longitude, latitude), note that the space in lat is important!!
    'type': 'asc2'  # Response format (e.g., asc2 for ASCII)
}
```

```python
# Make the GET request to the API
response = requests.get(base_url, params=params)
```

```python
# Check if the request was successful
if response.status_code == 200:
    # Print the response content
    print(response.text)
else:
    print(f"Error: {response.status_code}")
```
